# TAP Service Test Notebook

Automated tests for the Table Access Protocol (TAP) service.

## Imports

```python
from pyvo.dal.tap import TAPService
from astropy.table import Table
from lsst.rsp import RSPDiscovery
```

## Service Instantiation

```python
discovery = RSPDiscovery("dp2")
service = discovery.get_tap_client()
assert service is not None
print(f"TAP service URL: {service.baseurl}")
```

# Run Query

```python
query = """
SELECT * FROM TAP_SCHEMA.tables
"""
```

```python
for i in range(100):
    job = service.run_sync(query)
```

```python
assert len(job) > 40
```
