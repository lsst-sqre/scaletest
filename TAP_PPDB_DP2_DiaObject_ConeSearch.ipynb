# TAP Service Test Notebook

Automated tests for the Table Access Protocol (TAP) service.

## Imports

```python
from lsst.rsp import get_tap_service
from astropy.table import Table
from random import randrange, uniform
```

## Service Instantiation

```python
service = get_tap_service("ppdbtap")
assert service is not None
print(f"TAP service URL: {service.baseurl}")
```

```python
radius = 1.0
ra = uniform(226.2, 227.2)
dec = uniform(-40.8, -39.8)

print(ra, dec, radius)

str_center_coords = str(ra) + ", " + str(dec)
str_radius = str(radius)
```

## Asynchronous Query - DiaObject Variable Star Cone Search

```python
query = "SELECT ra, dec, diaObjectId, "\
        "r_psfFluxMean, nDiaSources, r_scienceFluxMean, r_scienceFluxMeanErr, "\
        "r_psfFluxNdata, r_psfFluxSigma "\
        "FROM dp2.DiaObject "\
        "WHERE CONTAINS(POINT('ICRS', ra, dec), "\
        "CIRCLE('ICRS', " + str_center_coords + ", " + str_radius + ")) = 1 "\
        "AND r_scienceFluxMean/r_scienceFluxMeanErr > 5 "\
        "AND nDiaSources > 100"
print(query)
results = service.run_async(query)
print(f"Retrieved {len(results)} rows")
```
