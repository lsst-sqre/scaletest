# TAP Service Test Notebook

Automated tests for the Table Access Protocol (TAP) service.

## Imports

```python
from pyvo.dal.tap import TAPService
from astropy.table import Table
import requests
from lsst.rsp import RSPDiscovery
```

## Service Instantiation

```python
discovery = RSPDiscovery("prompt")
service = discovery.get_tap_client()
assert service is not None
print(f"TAP service URL: {service.baseurl}")
```

# Run Query

```python
query = "SELECT TOP 1000000 * FROM ppdb.mpc_orbits"
job = service.submit_job(query)
job.run()
job.wait(phases=['COMPLETED', 'ERROR'], timeout=300)
print('Job phase is', job.phase)
print('Job ID is', job.job.jobid)
if job.phase == 'ERROR':
    job.raise_if_error()
```

```python
assert job.job.phase == "COMPLETED"
```
