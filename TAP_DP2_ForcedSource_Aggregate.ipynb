# TAP Service Test Notebook

Automated tests for the Table Access Protocol (TAP) service.

## Imports

```python
from lsst.rsp import get_tap_service
from astropy.table import Table
from random import randrange, uniform
```

## Service Instantiation

```python
service = get_tap_service("tap")
assert service is not None
print(f"TAP service URL: {service.baseurl}")
```

## Asynchronous Query - ForcedSource Full Scan Aggregate

```python
query = "SELECT band, COUNT(*) AS n, AVG(psfFlux) AS mean_flux, AVG(psfDiffFlux) AS mean_diff "\
        "FROM dp2.ForcedSource "\
        "WHERE psfFlux_flag = 0 "\
        "GROUP BY band"
print(query)
results = service.run_async(query)
print(f"Retrieved {len(results)} rows")
```
